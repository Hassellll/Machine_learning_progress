```python
#We import libraries

import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
#Load the dataset

data = pd.read_csv("Component_analysis_PCA.csv")
```

```python
data.head()
```

```python
#Separate features (X) and target variables (Y)

X = data.drop("Class", axis=1)
y = data["Class"]
```

```python
#Standardize the features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Apply PCA to reduce the datato two principal components

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
#Create a DataFrame with the transformed data

results = pd.DataFrame(
    X_pca,
    columns=["Principal Component 1", "Principal Component 2"]

)

results["Class"] = y
```

```python
#Display the explained Variance

print("Explained variance ratio of each component:")
print(pca.explained_variance_ratio_)

print("\nTotal explained variance:")
print(sum(pca.explained_variance_ratio_))

```

```python
#Plot the results

plt.figure(figsize=(8,6))

for label in results["Class"].unique():
    subset = results[results["Class"] == label]
    plt.scatter(
        subset["Principal Component 1"],
        subset["Principal Component 2"],
        label=label
    )
```

```python
plt.title("Principal Component Analysis (PCA)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(True)
plt.show()
```
