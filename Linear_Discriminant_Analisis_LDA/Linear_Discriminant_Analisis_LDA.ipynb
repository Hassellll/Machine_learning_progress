```python
#We import the libraries:

import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
#Load the CSV file

data = pd.read_csv("Linear_Discriminant_Analisis.csv")
```

```python
#Visualization

data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
#Define the feature variable (X)

X = data[["Age","Income","Experience"]]
#Define the target variable (Y)
Y = data["Class"]
```

```python
#Split the dataset into training and testing sets

x_train,x_test,y_train,y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
#Create the LDA model

lda = LinearDiscriminantAnalysis()

#train the model

lda.fit(x_train, y_train)
```

```python
#Make a prediction 

predictions = lda.predict(x_test)
```

```python
#Evaluate the model 

print("Accuracy:", accuracy_score(y_test, predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, predictions))

print("\nClassification Report:")
print(classification_report(y_test, predictions))
```

```python
#Predict a new observation

new_employee = [[38, 48000, 10]]
result = lda.predict(new_employee)
print("\nPredicted Class:", result[0])
```
