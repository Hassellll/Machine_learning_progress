```python
#Importamos librerias

import pandas as pd  #Libreria cientifica necesaria para trabajar con modelos
from sklearn.linear_model import LinearRegression #Funcion/algoritmo que nos ayudara a encontrar una relacion con predicciones
from sklearn.model_selection import train_test_split #Entrenamiento y prueba del modelo
from sklearn.metrics import mean_absolute_error #Metrica que nos dira el promedio de cuentas veces predijo mal el modelo
```

```python
#Colocamos el archivo y hacemos que pandas lo lea

data = pd.read_csv("ejercicio1.csv")
```

```python
#Visualizamos el contenido del dataset

data.head()
```

```python
data.info()
```

```python
#Usamos columns para no equivocarnos

data.columns
```

```python
#Definimos nuestras variables X y Y

X = data[["Clientes_por_hora"]] #Variables con datos que nos ayudaran a predecir
Y = data["Cafes_sevidos"] #Datos utiles y prediccion que haremos
```

```python
#Entrenamos al modelo:

x_train,x_test,y_train,y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
#Creamos al modelo

Model = LinearRegression()
Model.fit(x_train,y_train)
```

```python
#MAE
```

```python
#Predicion para los datos que tenemos y queremos predecir

prediccion1 = Model.predict(x_train)
prediccion2 = Model.predict(x_test)

print(prediccion1)
print(prediccion2)


```

```python
#promedio de error:

mean1 = mean_absolute_error(y_train, prediccion1)
mean2 = mean_absolute_error(y_test, prediccion2)

print(mean1)
print(mean2)
```

```python
#PASO 2
#  Elevar los errores al cuadrado


#¿Por qué hacemos esto? Porque si promediáramos directamente
#los errores, los positivos y negativos podrían cancelarse
#y dar una impresión engañosa del rendimiento del modelo.
```

```python
#PASO3 
#Calcular el promedio de los errores cuadrados
#3.1. Sumar los tres valores
#Para calcular un promedio, primero sumamos todos los valores:

#S= 100.000.000 + 100.000.000 + 400.000.000 
```

```python
MSE = np.mean(errores_al_cuadrado)

print(MSE)
```

```python
#¿Por qué aparece .0 al final? Porque NumPy devuelve normalmente un número decimal al calcular una media. El valor matemático es el mismo.
```

```python
#¿Qué significa?

#El modelo cometió errores de 10.000, 10.000 y 20.000 en los precios de las tres viviendas, considerando sus magnitudes.

#Al elevarlos al cuadrado y promediarlos, obtenemos 200.000.000.

#No significa que el modelo se equivoque por 200 millones de unidades monetarias. El MSE está expresado en unidades monetarias al cuadrado.
```
