```python
#Importamos librerias

import pandas as pd 
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
data = pd.read_csv("servicio_clientes.csv")
```

```python
#Definimos X y Y 

X = data[["edad","meses_cliente","soporte_mensual","retrasos_pago","uso_mensual_gb","contrato_anual"]]
Y = data["deserto"]
```

```python
#Creamos el modelo

model = LogisticRegression(max_iter=1000)
#entrenamos al modelo
cross = cross_val_score(model,X,Y, cv=5)
```

```python
mean = np.mean(cross)
```

```python
print(cross)
print("El promedio de el modelo es de:",mean)
```
